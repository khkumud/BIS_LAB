```python
import numpy as np


BOUNDS_LOW = np.array([5.0, 20.0, 0.5])
BOUNDS_HIGH = np.array([50.0, 200.0, 5.0])

def fitness_function(params):
    short_ma, long_ma, stop_loss = params
    if short_ma >= long_ma: return -99999


    simulated_profit = 100.0 / (abs(short_ma - 12) + abs(long_ma - 26) + 1)
    simulated_drawdown = 0.05 * (stop_loss)
    simulated_fees = 0.001 * (250 / (short_ma + 1))


    lambda1, lambda2 = 100.0, 50.0
    return simulated_profit - (lambda1 * simulated_drawdown) - (lambda2 * simulated_fees)

N, T = 15, 30  # 15 Particles, 30 Iterations
w, c1, c2 = 0.5, 1.5, 1.5


X = np.random.uniform(BOUNDS_LOW, BOUNDS_HIGH, (N, 3))
V = np.zeros((N, 3))
pbest = np.copy(X)
pbest_fit = np.array([fitness_function(p) for p in X])
gbest = pbest[np.argmax(pbest_fit)]
gbest_fit = np.max(pbest_fit)


for t in range(T):
    for i in range(N):
        fit = fitness_function(X[i])
        if fit > pbest_fit[i]:
            pbest[i] = X[i]
            pbest_fit[i] = fit
        if fit > gbest_fit:
            gbest = X[i]
            gbest_fit = fit


    r1, r2 = np.random.rand(N, 3), np.random.rand(N, 3)
    V = w*V + c1*r1*(pbest - X) + c2*r2*(gbest - X)
    X = np.clip(X + V, BOUNDS_LOW, BOUNDS_HIGH)


print("=== OPTIMIZATION COMPLETE ===")
print(f"Best Fitness Score: {gbest_fit:.4f}")
print(f"Optimal Short MA  : {round(gbest[0])} days")
print(f"Optimal Long MA   : {round(gbest[1])} days")
print(f"Optimal Stop Loss : {gbest[2]:.2f}%")
```
